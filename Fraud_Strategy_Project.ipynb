# Fraud Strategy Project

This project is organized as **four linked parts**. Open them in order:

1. [`notebooks/01_eda_and_rules.ipynb`](notebooks/01_eda_and_rules.ipynb) — EDA, candidate rules, **transform handoff**
2. [`notebooks/02_train_models.ipynb`](notebooks/02_train_models.ipynb) — train models, **model handoff**
3. [`notebooks/03_optimize_thresholds.ipynb`](notebooks/03_optimize_thresholds.ipynb) — validation optimization, **frozen strategy handoff**
4. [`notebooks/04_test_evaluation.ipynb`](notebooks/04_test_evaluation.ipynb) — untouched test evaluation

## Handoffs

| From → To | Artifact |
|---|---|
| Part 1 → Part 2 | `src/transform.py` |
| Part 2 → Part 3 | `artifacts/models/` + rules from Part 1 |
| Part 3 → Part 4 | `artifacts/strategy/frozen_strategy.json` |

Shared helpers live in `src/`. See `README.md` for the full layout.

The previous single-notebook version is archived at `notebooks/archive/Fraud_Strategy_Project_full.ipynb`.
